```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("student-mat.csv", sep=";")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicates Before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates After:", df.duplicated().sum())
```

```python
average_grade = df["G3"].mean()
students_above_15 = len(df[df["G3"] > 15])
correlation = df["studytime"].corr(df["G3"])
gender_average = df.groupby("sex")["G3"].mean()

print("Average Final Grade:", average_grade)

print("\nStudents scoring above 15:", students_above_15)

print("\nCorrelation between Study Time and Performance:")
print(correlation)

print("\nAverage Grades by Gender:")
print(gender_average)
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["G3"], bins=10, edgecolor="black")

plt.title("Distribution of Final Grades")
plt.xlabel("Grades")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["studytime"], df["G3"])

plt.title("Study Time vs Final Grades")
plt.xlabel("Study Time")
plt.ylabel("Final Grade")

plt.show()
```

```python
gender_average.plot(kind="bar")

plt.title("Male vs Female Average Score")
plt.xlabel("Gender")
plt.ylabel("Average Grade")

plt.show()
```
